### Описание признаков
##### Демографические данные

    CustomerID: Уникальный идентификатор клиента.
    BirthYear: Год рождения клиента.
    EducationLevel: Уровень образования клиента.
    MaritalStatus: Семейное положение клиента.
    AnnualIncome: Ежегодный доход домохозяйства клиента.
    NumChildren: Количество детей в семье клиента.
    NumTeenagers: Количество подростков в семье клиента.
    EnrollmentDate: Дата регистрации клиента в компании.

##### Поведение

    DaysSinceLastPurchase: Количество дней с момента последней покупки клиента.
    FiledComplaint: Указано, подавал ли клиент жалобы за последние 2 года (1 — да, 0 — нет).

##### Траты на продукты

    SpentOnWines: Сумма, потраченная на покупку вина за последние 2 года.
    SpentOnFruits: Сумма, потраченная на покупку фруктов за последние 2 года.
    SpentOnMeat: Сумма, потраченная на покупку мяса за последние 2 года.
    SpentOnFish: Сумма, потраченная на покупку рыбы за последние 2 года.
    SpentOnSweets: Сумма, потраченная на покупку сладостей за последние 2 года.
    SpentOnGold: Сумма, потраченная на покупку золота за последние 2 года.

##### Маркетинговые кампании

    DiscountedPurchases: Количество покупок со скидкой.
    AcceptedCampaign1: Указано, принял ли клиент предложение первой кампании (1 — да, 0 — нет).
    AcceptedCampaign2: Указано, принял ли клиент предложение второй кампании (1 — да, 0 — нет).
    AcceptedCampaign3: Указано, принял ли клиент предложение третьей кампании (1 — да, 0 — нет).
    AcceptedCampaign4: Указано, принял ли клиент предложение четвертой кампании (1 — да, 0 — нет).
    AcceptedCampaign5: Указано, принял ли клиент предложение пятой кампании (1 — да, 0 — нет).
    LastCampaignResponse: Указано, принял ли клиент предложение последней кампании (1 — да, 0 — нет).

##### Каналы покупок

    OnlinePurchases: Количество покупок, совершенных через веб-сайт компании.
    CatalogPurchases: Количество покупок, совершенных через каталог.
    InStorePurchases: Количество покупок, совершенных в магазине.
    WebsiteVisitsLastMonth: Количество визитов на веб-сайт компании за последний месяц.

##### Финансовые метрики

    ContactCost: Стоимость контакта с клиентом (постоянная метрика).
    RevenueGenerated: Доход, сгенерированный клиентом (постоянная метрика).

Оценим ассиметрию числовых данных

```python
skew_values = df[num_cols].skew()
print(f"{len(skew_values[skew_values < -1].index.tolist())} числовых признаков имеют левостороннюю (отрицательную) сильную асимметрию")
print(f"{len(skew_values[skew_values > 1].index.tolist())} числовых признаков имеют правостороннюю (положительную) сильную асимметрию")
```

```python
#Получим список признаков с правосторонней сильной асимметрией
num_cols_pos_skew = skew_values[skew_values > 1].index.tolist()
```

```python
def find_quantile_diff(df, quantile):
    """
         Вывод количества записей выше заданного квантиля.
         Возвращает входной датафрейм, отфильтрованный ниже заданного квантиля.
    """
    init_count = df.shape[0]
    for col in num_cols_pos_skew:
        limit = df[col].quantile(quantile)
        # Оставляем только те строки, которые не являются выбросами для данного признака
        df = df[df[col] <= limit]
    print(f"Выше {quantile} квантиля  {(init_count - df.shape[0]) * 100 / init_count:.2f}% записей")
    return df
```

```python
find_quantile_diff(df, 0.95)
find_quantile_diff(df, 0.99)
find_quantile_diff(df, 0.995)
df_cleaned = find_quantile_diff(df, 0.999)
```

Избавление от выбросов по 0.999 квантилю удалает 0.89% записей, что можно считать приемлемым. 

Оценим, как меняется асимметрия смещенных числовых признаков после очистки по 0.999 квантилю и после дополнительного преобразования (квадратный корень) для уменьшения смещенности 

```python
skew_before = df[num_cols_pos_skew].skew()
skew_after_cleaning = df_cleaned[num_cols_pos_skew].skew()
skew_after_sq_root_cleaning = (np.sqrt(df_cleaned[num_cols_pos_skew])).skew()

# Объединяем в одну таблицу оценки смещенности
skew_comparison = pd.DataFrame({
    'Смещенность до': skew_before,
    'Смещенность после очистки (0.995)': skew_after_cleaning,
    'Смещенность после очистки (0.995) + sqrt()': skew_after_sq_root_cleaning,
    'Улучшение': np.abs(skew_before) - np.abs(skew_after_sq_root_cleaning)
})

print(skew_comparison.sort_values(by='Улучшение', ascending = False).head(10).to_string())
```

По улучшениям можно увидеть, что очистка по 0.999 квантилю и дополнительному преобразованию сильно нормализует числовые признаки, что важно для K-means кластеризации.

Такое разделение может определять разные предпочтения в тратах и покупках, что может помочь в выделении кластеров (в бинарной кластеризации).

```python
def marital_group_status(status):
    if status in ['Married', 'Together']:
        return 'marital_partnered'
    else:
        return 'marital_single'
```

```python
df['marital_group'] = df['MaritalStatus'].apply(marital_group_status)
```

```python
def accepted_any_campaign(df):
    if (df['AcceptedCampaign1'] == 0) & (df['AcceptedCampaign2'] == 0) & (df['AcceptedCampaign3'] == 0) & \
    (df['AcceptedCampaign4'] == 0) & (df['AcceptedCampaign5'] == 0):
        return 1
    else:
        return 0
```

### 1.2. Подготовка данных для кластеризации

```python
# FiledComplaint', 'LastCampaignResponse', 'camp_ignored', 'is_parent' - числовые, но неявно категориальные признаки (значения 0 и 1)
num_columns_to_normalize = [c for c in df.columns if c not in \
                            ['EducationLevel', 'marital_group', 'FiledComplaint', 'LastCampaignResponse', 'camp_ignored', 'is_parent']]
```

```python
# проведем One-Hot Encoding для всех категориальных признаков
print(f"Размерность до One-Hot Encoding: {df.shape[1]} признаков")
df_encoded = pd.get_dummies(df, columns=['EducationLevel', 'marital_group'], dtype=int)
print(f"Размерность после One-Hot Encoding: {df_encoded.shape[1]} признаков")
```

```python
# проведем масштабирование числовых признаков
scaler = StandardScaler()
df_encoded[num_columns_to_normalize] = scaler.fit_transform(df[num_columns_to_normalize])
```

Для понижения размерности воспользуемся KernelPCA с подобраными параметрами для выделения кластеров.

## 2. Проведение кластеризации

```python
def visualize_clusters(df, clusters):
    """
        Провидит визуализацию кластеров через t-SNE
    """
    tsne = TSNE(n_components=2, random_state=42, n_jobs=-1, perplexity=50)
    X_tsne = tsne.fit_transform(df)
    
    df_vis = pd.DataFrame({
        'tSNE-1': X_tsne[:, 0],
        'tSNE-2': X_tsne[:, 1],
        'сluster': clusters,
    })

    plt.figure(figsize=(14, 10))
    
    # Цветовая схема: шум серый, кластеры — цвета по палитре
    palette_clusters = sns.color_palette("tab10", n_colors=len(set(clusters)) - (1 if -1 in clusters else 0))
    palette_clusters = {i: palette_clusters[i] for i in range(len(palette_clusters))}
    palette_clusters[-1] = '#95a5a6'  # Серый для шума
    
    scatter = sns.scatterplot(
        data=df_vis,
        x='tSNE-1',
        y='tSNE-2',
        hue='сluster',
        palette=palette_clusters,
        s=120,
        alpha=0.85,
        edgecolor='black',
        linewidth=0.5
    )
    
    plt.title("Кластеры, визуализированные через t-SNE", 
              fontsize=16, fontweight='bold', pad=20)
    plt.xlabel("t-SNE Component 1", fontsize=13, fontweight='bold')
    plt.ylabel("t-SNE Component 2", fontsize=13, fontweight='bold')
    
    # улучшенная легенда
    legend = plt.legend(
        title='Cluster ID', 
        title_fontsize=12,
        fontsize=11,
        loc='best',
        frameon=True,
        shadow=True
    )
    legend.get_frame().set_facecolor('white')
    legend.get_frame().set_alpha(0.9)
    
    plt.grid(True, alpha=0.3, linestyle='--', linewidth=0.7)
    plt.tight_layout()
    plt.show()
```

Проведем кластеризацию методом K-Means

```python
# определим оптимальное число кластеров, используя метод локтя
inertia = []
n = range(1, 15)

for i in n:
    km = KMeans(n_clusters=i,
                init='k-means++',
                n_init=10,
                max_iter=300,
                random_state=42)
    km.fit(df_encoded_PCA)
    inertia.append(km.inertia_)

plt.figure(figsize=(15, 9))
plt.plot(range(1, 15), inertia, marker='o')
plt.axvline(x=3, color='r', linestyle='--') 
plt.xlabel('Количество кластеров')
plt.ylabel('Инерция')
plt.tight_layout()
plt.show();
```

```python
# определим оптимальное число кластеров, используя метод силуэтов
score = []
k = range(2, 16)

for k_i in k:
    km = KMeans(n_clusters=k_i,
                init='k-means++',
                n_init=10,
                max_iter=300,
                tol=1e-04,
               random_state=42)
    km.fit(df_encoded_PCA)
    score.append(silhouette_score(df_encoded_PCA, km.labels_))
    
plt.plot(k, score)
plt.axvline(x=3, color='r', linestyle='--') 
plt.xlabel('Количество кластеров')
plt.ylabel('Коэффициент силуэта')
plt.show() 
```

Для метода K-Means оптимальным количеством кластеров будет 3. 

```python
n_clusters=3
km = KMeans(n_clusters=n_clusters,
            init='k-means++',
            n_init=10,
            max_iter=300,
            tol=1e-04,
            random_state=42)


km.fit_predict(df_encoded_PCA)
kmeans_clusters = km.labels_

score = silhouette_score(df_encoded_PCA,  km.labels_)
print(f"Качество кластеризации по методу силуэтов: {score:.2f}")

df_encoded_PCA_copy = df_encoded_PCA.copy()
# Добавляем кластеры во временный датафрейм для визуализации
df_encoded_PCA_copy['Cluster'] = km.labels_

visualize_clusters(df_encoded_PCA_copy,  km.labels_)
```

Проведем кластеризацию методом Agglomerative Clustering

```python
# определим оптимальное число кластеров, используя график расстояний
# вычисляем матрицу связей
linked = linkage(df_encoded_PCA, method='ward')

# визуализация дендрограммы
plt.figure(figsize=(10, 7))
dendrogram(linked,
            orientation='top',
            distance_sort='descending',
            show_leaf_counts=True)

# добавим линию разреза для отделения кластеров
plt.axhline(y=8, color='r', linestyle='--') 
plt.title('Дендрограмма иерархической кластеризации')
plt.ylabel('Расстояние (Ward distance)')
plt.show()
```

```python
# определим оптимальное число кластеров, используя метод силуэтов
score = []
k = range(2, 16)

for k_i in k:
    agglomerative = AgglomerativeClustering(
        n_clusters=k_i, linkage='ward')
    agglomerative.fit(df_encoded_PCA)
    score.append(silhouette_score(df_encoded_PCA, agglomerative.labels_))
    
plt.plot(k, score)
plt.axvline(x=3, color='r', linestyle='--') 
plt.xlabel('Количество кластеров')
plt.ylabel('Коэффициент силуэта')
plt.show() 
```

Для метода Agglomerative Clustering оптимальным количеством кластеров будет 3. 

```python
n_clusters=3
agglomerative = AgglomerativeClustering(
    n_clusters=n_clusters, linkage='ward')
agglomerative.fit(df_encoded_PCA)

df_encoded_y = agglomerative.fit_predict(df_encoded_PCA)

score = silhouette_score(df_encoded_PCA,  df_encoded_y)
print(f"Качество кластеризации по методу силуэтов: {score:.2f}")

df_encoded_PCA_copy = df_encoded_PCA.copy()
# Добавляем кластеры во временный датафрейм для визуализации
df_encoded_PCA_copy['Cluster'] = df_encoded_y

visualize_clusters(df_encoded_PCA_copy, df_encoded_y)
```

```python
spent_features_to_divide = ['SpentOnWines', 'SpentOnFruits', 'SpentOnMeat', 'SpentOnFish', 'SpentOnSweets', 'SpentOnGold']
# доли годового дохода на траты на продукты
spent_partition_names = ['income_percent_' + col for col in spent_features_to_divide]

# для признаков 'AnnualIncome', 'SpentOnWines', 'SpentOnFruits', 'SpentOnMeat', 'SpentOnFish', 'SpentOnSweets', 'SpentOnGold'
# проводилось преобразование sqrt()
spent_features_to_divide_sq_df = df_orig[spent_features_to_divide]**2
annual_income_sq_df = df_orig['AnnualIncome']**2

df_orig[spent_partition_names] = spent_features_to_divide_sq_df.div(annual_income_sq_df, axis=0).mul(100)
# средние доли годового дохода для конкретных траты на продукты в разрезе кластеров
spent_partition_means_per_cluster = df_orig[spent_partition_names + ['Cluster']].groupby('Cluster').mean().T

ax = spent_partition_means_per_cluster.plot(kind='barh')
ax.set_xlabel("Доля от годового дохода, в %");
new_labels = ['Золото', 'Сладости', 'Рыба', 'Мясо', 'Фрукты', 'Вино']
ax.set_yticklabels(new_labels, rotation=0);
```

Представители кластера 0 сравнительно больше тратят на продукты.\
Представители кластера 1 сравнительно меньше тратят на продукты.\
Представители кластера 0 и 2 относительно годового дохода много тратят на золото.

```python
spent_features = ['SpentOnWines', 'SpentOnFruits', 'SpentOnMeat', 'SpentOnFish', 'SpentOnSweets', 'SpentOnGold']
spent_features_df = df_orig[['Cluster'] + spent_features]
# для признаков SpentOnWines', 'SpentOnFruits', 'SpentOnMeat', 'SpentOnFish', 'SpentOnSweets', 'SpentOnGold'
# проводилось преобразование sqrt()
spent_features_df[spent_features] = spent_features_df[spent_features]**2
df_spent = spent_features_df.groupby('Cluster')[spent_features].mean()

# расчет масштаба радиусов
sums = df_spent.sum(axis=1)
# нормируем радиусы
max_radius = 1.0
radii = np.sqrt(sums) / np.sqrt(sums.max()) * max_radius

# настройка сетки
n_rows = len(df_spent)
cols = 2
rows = math.ceil(n_rows / cols)
fig, axes = plt.subplots(rows, cols, figsize=(10, 5 * rows))
axes = axes.flatten()

colors = plt.get_cmap('tab10').colors

# построение
for i, (idx, row) in enumerate(df_spent.iterrows()):
    # рисуем диаграмму с индивидуальным радиусом
    wedges, texts, autotexts = axes[i].pie(
        row, 
        radius=radii[i],
        autopct='%1.1f%%',
        pctdistance=0.85 if radii[i] > 0.5 else 1.2, # сдвигаем цифры, если круг мелкий
        startangle=90,
        colors=colors
    )
    
    # добавляем название и общую сумму для наглядности
    axes[i].set_title(f"Кластер {idx}\n(Средняя сумма трат: {sums[i]:.1f})", pad=20, fontweight='bold')

# легенда 
fig.legend(wedges, ['Вино', 'Фрукты', 'Мясо', 'Рыба', 'Сладости', 'Золото'], 
           title="Траты", loc="center right", bbox_to_anchor=(1.1, 0.5))

for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show();
```

Клиенты из кластера 0 тратят больше всего на вино (45.8%) и мясо (31.0%). По мясу имеют наибольшую долю в тратах относительно других кластеров.\
Клиенты из кластера 1 отличаются наличием наибольшей доли золота (17.9%) в расходах. Но средние расходы на порядки меньше, чем в других кластерах.\

&emsp;&emsp; В кластере 2 представлены клиенты со сравнительно малым доходом, с подростками, с высоким уровнем образования, младших и старших возрастов, участвующие в маркетинговых кампаниях на среднем уровне, покупающие часто в магазинах, активно покупающие со скидками, при этом тратят на среднем уровне. Из покупок примерно 2/3 трат составляет вино.

## 4. Формирование маркетинговых рекомендаций

&emsp;&emsp; Возможные предложения для клиентов (в порядке убывания потенциального эффекта):
* специальные предложения на покупки вина (в первую очередь) и мяса (во вторую очередь) в магазинах
* специальные предложения для клиентов с, как минимум, высшим образованием
* скидки на покупку золота для клиентов с базовым образованием или с детьми/подростками
* скидки для клиентов с подростками, особенно в случае покупки в магазинах

&emsp;&emsp; Клиенты кластеров 0 и 2 имеют общие характеристики - сравнительно высокий доход, активно участвуют в маркетинговых кампаниях, покупают в магазинах, покупают очень много вина и мяса, с Graduation, PhD и Master уровнем образования, поэтому спец. предложения может увеличить продажи.

&emsp;&emsp; Клиенты кластеров 2 выделяются наличием подростков и склонностью к скидкам, поэтому скидки в магазинах могут привлечь таких клиентов.

&emsp;&emsp; У клиентов кластера 1 средний доход, но на порядки более низкие траты, сравнивая с иными кластерами, существенная доля трат на золото (18 %), наличие детей свидетельствует о сберегательном поведении, а характерное базовое образование может указывать на слабую финансовую грамотность. Для привлечения таких клиентов подходящим вариантом будут скидки в целом и, особенно, на золото, но в силу сравнительно низких трат этой группы, усилия для привлечения стоит считать низкоприоритетными.
