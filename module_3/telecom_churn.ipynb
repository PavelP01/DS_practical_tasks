```python
import pandas as pd
import numpy as np
from scipy import stats

def num_cols_stat_report(df):
    """
        Создает отчет по KS-статистике (D), асимметрии (смещенность) и выбросам для числовых признаков
    """
    report = []
    numeric_cols = df.columns

    for col in numeric_cols:
        data = df[col].dropna()
        # Стандартизация для теста Колмогорова-Смирнова
        standardized_data = (data - data.mean()) / data.std()
        
        # Проведение теста Колмогорова-Смирнова
        ks_stat, _ = stats.kstest(standardized_data, 'norm')
        
        # Оценка асимметрии 
        skew = data.skew()
        
        # Оценка процент выбросов (по правилу 3-х сигм)
        outliers_count = ((data < data.mean() - 3 * data.std()) | 
                          (data > data.mean() + 3 * data.std())).sum()
        outliers_percents = (outliers_count / len(data)) * 100
        
        report.append({
            'Признак': col,
            'KS-статистика (D)': round(ks_stat, 4),
            'Асимметрия (смещенность)': round(skew, 3),
            'Выбросы % (>3 сигм)': round(outliers_percents, 2)
        })
    
    return pd.DataFrame(report).sort_values(by='Выбросы % (>3 сигм)', ascending=False)
```

```python
report = num_cols_stat_report(num_cols_copy)
```

```python
def print_num_cols_summary_stats(report_df):
    """
        Вывод компонентов отчета по KS-статистике (D), асимметрии (смещенность) и выбросам для числовых признаков
    """
    total_features = len(report_df)
    
    # Расчет количества "проблемных" признаков
    ks_attention = (report_df['KS-статистика (D)'] > 0.1).sum()
    skew_attention = (report_df['Асимметрия (смещенность)'].abs() > 1).sum()
    outlier_attention = (report_df['Выбросы % (>3 сигм)'] > 0.27).sum()
    
    # Вывод в процентах
    print(f"Статистика по датасету ({total_features} числовых признаков):")
    print(f"1. Отклонение по тесту Колмогорова-Смирнова (> 0.1): {(ks_attention/total_features)*100:.1f}% признаков")
    print(f"2. Высокая асимметрия (|Skew| > 1): {(skew_attention/total_features)*100:.1f}% признаков")
    print(f"3. Избыток выбросов (> 0.27%, за пределами \u03bc \u00b1 3\u03c3): {(outlier_attention/total_features)*100:.1f}% признаков")
```

```python
print_num_cols_summary_stats(report)
```

Избавимся от выбросов.

```python
def find_quantile_diff(df, quantile):
    """
         Вывод количества записей выше заданного квантиля.
         Возвращает входной датафрейм, отфильтрованный ниже заданного квантиля.
    """
    init_count = df.shape[0]
    for col in num_cols_pos_skew:
        limit = df[col].quantile(quantile)
        # Оставляем только те строки, которые не являются выбросами для данного признака
        df = df[df[col] <= limit]
    print(f"Выше {quantile} квантиля  {(init_count - df.shape[0]) * 100 / init_count:.2f}% записей")
    return df
```

```python
find_quantile_diff(data, 0.95)
find_quantile_diff(data, 0.99)
find_quantile_diff(data, 0.995)
data_cleaned = find_quantile_diff(data, 0.999)
```

Избавление от выбросов по 0.999 квантилю удалает 6.45% записей, что можно считать приемлемым. \
Применение меньшего порога квантиля (<99.9 %) к удалению выбросов приведет к существенной потере данных вследствие сильной правосторонней асимметрии.

Оценим, как меняется асимметрия после очистки по 0.999 квантилю и после дополнительного преобразования для уменьшения смещенности 
$$ \text{sign}(x) \cdot |x|^{1/3} $$

```python
skew_before = data[num_cols_pos_skew].skew()
skew_after_cleaning = data_cleaned[num_cols_pos_skew].skew()
skew_after_cube_root_cleaning = (np.sign(data_cleaned[num_cols_pos_skew]) * np.power(np.abs(data_cleaned[num_cols_pos_skew]), 1/3)).skew()

# Объединяем в одну таблицу оценки смещенности
skew_comparison = pd.DataFrame({
    'Смещенность до': skew_before,
    'Смещенность после очистки (0.999)': skew_after_cleaning,
    'Смещенность после очистки (0.999) + доп. преобразование': skew_after_cube_root_cleaning,
    'Улучшение': np.abs(skew_before) - np.abs(skew_after_cube_root_cleaning)
})

print(skew_comparison.sort_values(by='Улучшение', ascending = False).head(10).to_string())
```

По наибольшим улучшениям можно увидеть, что очистка по 0.999 квантилю и дополнительному преобразованию сильно нормализует признаки.

```python
skew_values_before = data[num_cols_pos_skew].skew().sort_values()
skew_values_after_cleaning = data_cleaned[num_cols_pos_skew].skew().sort_values()
skew_values_after_cube_root_cleaning = \
(np.sign(data_cleaned[num_cols_pos_skew]) * np.power(np.abs(data_cleaned[num_cols_pos_skew]), 1/3)).skew()

# Создаем общий DataFrame для визуализации
df_compare = pd.DataFrame({
    'До': skew_values_before,
    'Очистка (0.999)': skew_values_after_cleaning,
    'Очистка (0.999) + доп. преоб-ие': skew_values_after_cube_root_cleaning,
}).melt(var_name='Варианты', value_name='Смещенность')

plt.figure(figsize=(10, 6))

# Рисуем боксплот
ax = sns.boxplot(data=df_compare, x='Смещенность', y='Варианты', palette='Set2')

# Устанавливаем симметричную логарифмическую шкалу
ax.set_xscale('symlog')

# Добавляем вертикальные линии порогов
plt.axvline(1, color='red', linestyle='--', alpha=0.5, label='Порог положительной смещенности')
plt.axvline(-1, color='red', linestyle='-.', alpha=0.5, label='Порог отрицательной смещенности')
plt.axvline(0, color='black', linewidth=0.8)

plt.title('Сравнение асимметрии (Symlog шкала)')
plt.grid(True, which="both", ls="-", alpha=0.2)
plt.legend()
plt.show()
```

```python
def highlight_high_corr(val):
    """
        Служит для подсветки ячеек таблиц, соответствующих высокому коэффициенту корреляции (> 0.9)
    """
    color = '#FFD700' if 0.9 <= abs(val) < 1.0 else ''
    return f'background-color: {color}'
```

Оценим визуально пары признаков с сильной корреляцией

```python
#Считаем корреляцию
corr = data.corr()

#Оставляем только признаки, у которых есть сильные связи 
features_to_keep = corr.columns[(corr.abs() >= 0.9).sum() > 1]
filtered_corr = corr.loc[features_to_keep, features_to_keep]

# Скрываем верхний треугольник (np.triu) и значения, которые по модулю меньше 0.9
mask = np.triu(np.ones_like(filtered_corr, dtype=bool)) | (filtered_corr.abs() < 0.9)

plt.figure(figsize=(12, 10))

sns.heatmap(
    filtered_corr, 
    mask=mask, 
    annot=False, 
    fmt=".2f",
    cmap='Reds', 
    cbar=True,
    square=True,
    linewidths=.5,    
    annot_kws={"size": 10}
)

plt.title('Пары признаков с корреляцией >= 0.9', fontsize=15)
plt.show()
```

Проведем удаление признаков в каждой группы признаков, \
оставляя один из коррелирующих признаков, если все элементы сильно коррелируют, или\
оставляя один или более признаков, если часть элементов сильно коррелируют \
(т.е. если A сильно связан с B, B сильно связан с C, но A и C связаны слабо, то удалим только B, оставив A и C)

```python
corr = data[['unan_vce_Mean', 'plcd_vce_Mean']].corr()
display(corr.style.applymap(highlight_high_corr))
print("Удалим признак unan_vce_Mean")
data.drop(['unan_vce_Mean'], axis = 1, inplace = True)
```

## 3.Построение модели классификации

```python
from sklearn.model_selection import train_test_split

X = data.drop(['churn'], axis=1)
y = data['churn']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=21, stratify=y
)
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
#проведем скалирование признаков для тренировочных и тестовых данных
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
#проведем скалирование числовых признаков для всех данных (для кросс-валидации)
from sklearn.preprocessing import MinMaxScaler
all_scaler = MinMaxScaler()

X_scaled = all_scaler.fit_transform(X)
```

```python
def create_quality_report(actual, prediction, label = 'metrics'):
    """
        Вывод значений метрик accuracy, precision, recall, F1-score
    """
    acs = accuracy_score(actual, prediction)
    prs = precision_score(actual, prediction)
    rs = recall_score(actual, prediction)
    f1 = f1_score(actual, prediction)
    return pd.DataFrame([acs, prs, rs, f1], index=['accuracy', 'precision', 'recall', 'f1_score'], columns=[label])
```

```python
def display_model_overall_report(train_report, test_report):
    """
        Вывод значений метрик accuracy, precision, recall, F1-score.
        Желтым выделяется максимальное значение среди значений метрики.
    """
    overall_report = pd.concat([train_report, test_report], axis=1)
    a = overall_report.style\
        .highlight_max(color='yellow', axis=1)\
        .set_caption("Качество обучения")
    display(a)
```

```python
from sklearn.metrics import roc_auc_score, roc_curve

def plot_roc_curve(actual, prob_prediction):
    """
        Формирует изображение для roc кривой
    """
    fpr, tpr, thresholds = roc_curve(actual, prob_prediction)
    auc_score = roc_auc_score(actual, prob_prediction)
    
    plt.plot(fpr, tpr, label='ROC curve ')
    plt.plot([0, 1], [0, 1])
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('ROC AUC: {:.3f}'.format(auc_score))
    plt.show()
```

### Кросс-валидация и оценка качества

Проведем кросс-валидацию моделей классификации

```python
from sklearn.model_selection import cross_validate

scoring = ['accuracy', 'precision', 'recall', 'f1', 'roc_auc']
```

```python
#Подготовка
log_reg = LogisticRegression(n_jobs=-1)
svc = SVC(C=10, cache_size = 1500)
rnd_clf = RandomForestClassifier(max_leaf_nodes=16, random_state=42, n_jobs=-1)
ada_clf = AdaBoostClassifier(
    DecisionTreeClassifier(max_depth=1),
    n_estimators=100,
    learning_rate=0.5,
    random_state=42)

predictors = {
    "Logistic Regression": log_reg,
    "SVC": svc,
    "Random Forest": rnd_clf,
    "Ada boosting": ada_clf
}

results_list = []

#Сбор данных
for name, model in predictors.items():
    cv_results = cross_validate(model, X_scaled, y, cv=5, scoring=scoring, n_jobs=-1)
    
    model_scores = {
        "Модель": name,
        "Время выполнения, c": cv_results['fit_time'].mean() # Добавим время обучения
    }
    for metric in scoring:
        model_scores[metric.capitalize()] = cv_results[f'test_{metric}'].mean()
    
    results_list.append(model_scores)

df_results = pd.DataFrame(results_list).set_index("Модель")

#Функции стилизации
def highlight_max(s):
    """Подсветка максимума (для метрик качества)"""
    return ['background-color: #2ecc71; font-weight: bold; color: black' if v == s.max() else '' for v in s]

def highlight_min(s):
    """Подсветка минимума (для времени обучения)"""
    return ['background-color: #f1c40f; font-weight: bold; color: black' if v == s.min() else '' for v in s]

styled_df = df_results.style \
    .apply(highlight_max, subset=df_results.columns[1:]) \
    .apply(highlight_min, subset=['Время выполнения, c']) \
    .format("{:.4f}")

styled_df
```

Время выполнения: сколько секунд в среднем уходит на обучение одного фолда

```python
#Анализ лучших моделей
metrics_only = df_results.drop(columns=['Время выполнения, c'])
best_per_metric = metrics_only.idxmax()
overall_winner = best_per_metric.value_counts().idxmax()

for metric, model in best_per_metric.items():
    print(f"Лучшая модель по {metric}: {model} ({df_results.loc[model, metric]:.4f})")

print(f"\nПо большинству метрик лучше подходит: {overall_winner}")
```

Логистическая регрессия уступает незначительно SVC по roc_auc и precision, при этом выигрывает сильно по recall и f1 метрикам.

## 4. Тонкая настройка модели  

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression
```

```python
# Логистическая регрессия
param_grid = {
    'penalty': ['l1', 'l2', 'elasticnet', 'none'],
    'C': [0.1, 1, 10],
    'solver': ['liblinear', 'saga', 'lbfgs'], 
    'max_iter': [100, 200, 500],
}
grid_search = GridSearchCV(
    LogisticRegression(n_jobs=-1),
    param_grid,
    cv=5,                       
    scoring='roc_auc',
    n_jobs=-1,
    verbose=1
)
grid_search.fit(X_scaled, y)

print("Наилучшие параметры модели:", grid_search.best_params_)

best_model = grid_search.best_estimator_

train_report = create_quality_report(y_train, best_model.predict(X_train_scaled), 'тренир. данные')
test_report = create_quality_report(y_test, best_model.predict(X_test_scaled), 'тест. данные')
display_model_overall_report(train_report, test_report)

plot_roc_curve(y_test, best_model.predict_proba(X_test_scaled)[:,1])
```
